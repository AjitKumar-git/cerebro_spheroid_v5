## Step 4: Autonomous Generation & Multi-Shell Deliberation Probe
Test pure autonomous generation without shortcut bypass dependencies.

```python
# ==============================================================================
# 💬 STEP 4: INTERACTIVE PROMPT TESTING & MULTI-SHELL PROBING
# ==============================================================================
import os, jax
import jax.numpy as jnp
import numpy as np
from src.jax.v5_model import ModelConfigV5, PRESETS_V5, training_rollout
from src.jax.checkpoint import load_torch_pt_to_jax
from src.cloud_sync import pull_checkpoint, get_sync_token
from src.corpus import build_tokenizer, format_stimulus_prompt, clean_motor_response, encode_no_special

CKPT_PATH = "/content/checkpoints/recovery/v5_small_recovery_best.pt"
if not os.path.exists(CKPT_PATH):
    CKPT_PATH = "/content/checkpoints/recovery/v5_small_recovery_latest.pt"
if not os.path.exists(CKPT_PATH):
    CKPT_PATH = "checkpoints/v5_small_latest.pt"
if not os.path.exists(CKPT_PATH):
    print("🌐 Pulling latest checkpoint from bioflo.in...")
    pull_checkpoint(CKPT_PATH, target_type="latest", model="v5_small_recovery")

print(f"Loading checkpoint: {CKPT_PATH}...")
tokenizer = build_tokenizer("llama")

cfg_dict = PRESETS_V5["v5_small"].copy()
cfg_dict["thinking_cycles"] = 4
cfg_dict["use_adaptive_thinking"] = False
cfg_dict["sensory_position_steps"] = 4096
cfg_dict["max_phase_steps"] = 4096
cfg_dict["max_seq_len"] = 4096
config = ModelConfigV5(**cfg_dict)

params, buffers, meta = load_torch_pt_to_jax(CKPT_PATH, config, target_dtype=jnp.bfloat16)
print(f"✓ Loaded model step {meta.get('step', 0):,} (Loss: {meta.get('loss', 0.0):.4f})")

def probe_generation(prompt_text: str, max_tokens: int = 64) -> str:
    stimulus = format_stimulus_prompt(prompt_text)
    tokens = encode_no_special(tokenizer, stimulus)
    prompt_arr = jnp.array([tokens], dtype=jnp.int32)
    
    out_logits = training_rollout(
        params=params,
        prompts=prompt_arr,
        target_len=max_tokens,
        buffers=buffers,
        config=config,
        targets=None,
        feedback_tokens=None,
        pure_teacher_forcing=False,
        deliberation_mode="teacher_forcing",
        repetition_penalty=1.15,
    )
    gen_token_ids = np.array(jnp.argmax(out_logits, axis=-1)[0])
    raw = tokenizer.decode(gen_token_ids, skip_special_tokens=False)
    return clean_motor_response(raw).strip()

for prompt in [
    "The fundamental principles of thermodynamics govern",
    "Explain the concept of neural deliberation in biological systems:",
]:
    print(f"\n💡 Prompt:   {prompt}")
    print(f"🔊 Generated: {probe_generation(prompt)}")
```
