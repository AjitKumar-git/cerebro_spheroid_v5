## Step 2: Configure & Launch Training
Launches pretraining for `v5_small` (105.9M parameters, 16 shells, 40 shortcut tracts).
- **Hardware Auto-Tuning**: Micro-batches and remat policy are automatically calibrated for optimal TPU MXU throughput.
- **From Scratch**: Starts with clean random initialization (seed 42) across full 4,096-token sequence windows.
- **New Rectified Dataset**: 85% FineWeb (sample-10BT) broad web + 15% Code (The Stack v2) with 100% loss-bearing tokens.
- **Remote Sync**: Checkpoints (`latest.pt`, `best.pt`) and live metrics stream directly to `bioflo.in`.

```python
# ==============================================================================
# ⚡ Step 2: Configure & Launch V5 Small Pretraining
# ==============================================================================
from training.train_v5_tpu_jax import run_training

# ── Primary Configuration ─────────────────────────────────────────────────────
MODEL_PRESET       = 'v5_small'               # 105.9M Parameters (16 Shells, 40 Shortcut Tracts)
FROM_SCRATCH       = False                     # True: Fresh start at Step 0 with random seed 42
AUTO_TUNE          = True                     # Auto-tune discovers micro-batch & remat on TPU v5e-1
EFFECTIVE_BATCH    = 16                       # Global effective batch size (65,536 tok/step at 4096 ctx)
TARGET_TOKENS      = 15_700_000_000          # 15.7 Billion Loss-Bearing Token Target Budget (~60,000 steps)
DATASET_SOURCE     = 'pretrain'               # Rectified 85% FineWeb (sample-10BT) + 15% Code (The Stack v2)

FORCE_TUNE = False

# ── Execute High-Throughput JAX Runner ─────────────────────────────────────────
run_training(
    preset=MODEL_PRESET,
    from_scratch=FROM_SCRATCH,
    auto_tune=AUTO_TUNE,
    force_tune=FORCE_TUNE,                    # Re-run all batch-size candidates instead of using a cached winner
    effective_batch_size=EFFECTIVE_BATCH,
    target_tokens=TARGET_TOKENS,
    dataset_source=DATASET_SOURCE,
    total_window=4096,
    sensory_position_steps=4096,
    max_phase_steps=4096,
    checkpoint_dir="/content/checkpoints/v5_small_scratch",
    cloud_sync=True,
)
```

## Step 4: Interactive Live Inference & Prompt Testing
Generate completions from your trained checkpoint (locally cached or auto-pulled from bioflo.in).

```python
# ==============================================================================
# 💬 Step 4: Interactive Live Inference & Prompt Testing
# ==============================================================================
import os, jax
import jax.numpy as jnp
import numpy as np
from src.jax.v5_model import ModelConfigV5, PRESETS_V5, training_rollout
from src.jax.checkpoint import load_torch_pt_to_jax
from src.cloud_sync import pull_checkpoint
from src.corpus import build_tokenizer, format_stimulus_prompt, clean_motor_response, encode_no_special

CKPT_PATH = "/content/checkpoints/v5_small_scratch/v5_small_scratch_best.pt"
if not os.path.exists(CKPT_PATH):
    CKPT_PATH = "/content/checkpoints/v5_small_scratch/v5_small_scratch_latest.pt"
if not os.path.exists(CKPT_PATH):
    print("🌐 Checkpoint not found locally. Pulling best.pt from bioflo.in...")
    pull_checkpoint(CKPT_PATH, target_type="best", model="v5_small")

print(f"Loading checkpoint: {CKPT_PATH}...")
tokenizer = build_tokenizer("llama")

cfg_dict = PRESETS_V5["v5_small"].copy()
cfg_dict["sensory_position_steps"] = 4096
cfg_dict["max_phase_steps"] = 4096
cfg_dict["max_seq_len"] = 4096
config = ModelConfigV5(**cfg_dict)

params, buffers, meta = load_torch_pt_to_jax(CKPT_PATH, config, target_dtype=jnp.bfloat16)
print(f"✓ Loaded model from Step {meta.get('step', 0):,} (Loss: {meta.get('loss', 0.0):.4f})")

def generate(prompt_text: str, max_tokens: int = 128) -> str:
    stimulus = format_stimulus_prompt(prompt_text)
    tokens = encode_no_special(tokenizer, stimulus)
    prompt_arr = jnp.array([tokens], dtype=jnp.int32)
    
    out_logits = training_rollout(
        params=params,
        prompts=prompt_arr,
        target_len=max_tokens,
        buffers=buffers,
        config=config,
        targets=None,
        feedback_tokens=None,
        pure_teacher_forcing=False,
        deliberation_mode="teacher_forcing",
        repetition_penalty=1.15,
    )
    gen_token_ids = np.array(jnp.argmax(out_logits, axis=-1)[0])
    raw = tokenizer.decode(gen_token_ids, skip_special_tokens=False)
    return clean_motor_response(raw).strip()

test_prompt = "What is the capital of France?"
print(f"\nPrompt:   {test_prompt}")
print(f"Response: {generate(test_prompt)}")
```
